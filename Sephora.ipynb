```python
import pandas as pd
product_df = pd.read_csv("/content/product_info.csv")
```

```python
product_df.head(5)
```

```python
print(product_df.columns)
print(product_df.shape)
print(product_df.size)
```

```python
product_df.drop(['brand_id', 'size', 'variation_type', 'variation_value', 'variation_desc', 'ingredients', 'value_price_usd', 'sale_price_usd', 'highlights', 'child_count', 'child_max_price', 'child_min_price'], axis = 1, inplace = True)
```

```python
print(product_df.shape)
```

```python
product_df.isnull().sum()
```

```python
product_df.dropna(subset = ['rating','reviews'], inplace = True)
product_df['secondary_category'].fillna('Unknown', inplace = True)
```

```python
product_df.isnull().sum()
```

```python
product_df.shape
```

```python
review_0_df = pd.read_csv('/content/reviews_0-250.csv')
review_1_df = pd.read_csv('/content/reviews_1250-end.csv')
review_2_df = pd.read_csv('/content/reviews_250-500.csv')
review_3_df = pd.read_csv('/content/reviews_500-750.csv')
review_4_df = pd.read_csv('/content/reviews_750-1250.csv')

reviews_df = pd.concat([review_0_df,review_1_df,review_2_df,review_3_df,review_4_df])
```

```python
reviews_df.shape
```

```python
reviews_df.columns
```

```python
reviews_df = reviews_df.drop(["Unnamed: 0", "author_id", "total_neg_feedback_count", "total_pos_feedback_count", "review_text", "eye_color", "hair_color", "product_name", "brand_name", "price_usd"], axis = 1)
```

```python
reviews_df.columns
```

```python
reviews_df.isnull().sum()
```

```python
reviews_df = reviews_df.drop(['helpfulness'], axis = 1)
reviews_df['skin_type'] = reviews_df['skin_type'].fillna('Unknown')
```

```python
reviews_df.shape
```

```python
reviews_df.isnull().sum()
```

```python
reviews_df['submission_time'] = pd.to_datetime(reviews_df['submission_time'])
reviews_df['submission_time'].dtype
```

```python
reviews_df['submission_time']
```

```python
product_df.to_csv('clean_products.csv', index = False)
reviews_df.to_csv('clean_reviews.csv', index = False)
```
